# Chapter 12: Conditional Statements

## Definition

Conditional statements are used to make decisions based on conditions.

## if Statement

The `if` statement executes a block of code only when the condition is True.

```python
age = 20

if age >= 18:
    print("Eligible to Vote")
```

## if...else Statement

The `if...else` statement executes one block if the condition is True and another block if it is False.

```python
number = 15

if number % 2 == 0:
    print("Even")
else:
    print("Odd")
```

## if...elif...else Statement

The `elif` statement checks multiple conditions one by one.

```python
marks = 85

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
else:
    print("Grade C")
```

## Nested if Statement

A nested `if` means an `if` statement inside another `if` statement.

```python
age = 25
citizen = True

if age >= 18:
    if citizen:
        print("Eligible to Vote")
```

## Summary

- if
- if...else
- if...elif...else
- Nested if

# Chapter 14: Loop Control Statements

## Definition

Loop Control Statements are used to control the execution of loops.

## Types

1. break
2. continue
3. pass

```python
# break Statement

for i in range(1, 6):
    if i == 4:
        break
    print(i)
```

## continue Statement

The `continue` statement skips the current iteration and moves to the next iteration.

```python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

## pass Statement

The `pass` statement does nothing. It is used as a placeholder.

```python
for i in range(1, 6):
    if i == 3:
        pass
    print(i)
```

## break in while Loop

```python
count = 1

while True:
    if count == 5:
        break
    print(count)
    count += 1
```

## continue in while Loop

```python
count = 0

while count < 5:
    count += 1
    if count == 3:
        continue
    print(count)
```

## Summary

- break
- continue
- pass
- break with while Loop
- continue with while Loop

Example:

list[start : end : step]

```python
fruits = ("Apple", "Banana", "Mango", "Orange")

print(fruits[0])
print(fruits[2])
print(fruits[-1])
print(fruits[-2])
```

## Tuple Slicing

Slicing is used to access a portion of a tuple.

### Syntax

tuple[start : end : step]

```python
numbers = (10, 20, 30, 40, 50, 60)

print(numbers[1:4])
print(numbers[:3])
print(numbers[3:])
print(numbers[::2])
print(numbers[::-1])
```

# Chapter 18: Sets

## Definition

A **Set** is an unordered, mutable, and unindexed collection in Python. It stores only unique values, which means duplicate values are automatically removed.

Sets are created using curly braces `{}` or the `set()` function.

Example:
- `{10, 20, 30}`
- `{"Apple", "Banana", "Mango"}`
- `{1, "Python", 3.14, True}`

```python
# Creating a Set

numbers = {10, 20, 30, 40, 50}
fruits = {"Apple", "Banana", "Mango"}
mixed = {10, "Python", 3.14, True}

print(numbers)
print(fruits)
print(mixed)
```

## Duplicate Values

Sets automatically remove duplicate values.

# Chapter 19: Dictionaries

## Definition

A **Dictionary** is an unordered, mutable collection of data that stores values in **key-value pairs**.

Each key must be unique, and it is used to access its corresponding value.

Dictionaries are created using curly braces `{}`.

Example:
- `{"name": "Ankit", "age": 21}`
- `{"id": 101, "course": "Python"}`

```python
# Creating a Dictionary

student = {
    "name": "Ankit",
    "age": 21,
    "city": "Delhi"
}

print(student)
```

## Accessing Values

Dictionary values are accessed using their keys.

```python
student = {
    "name": "Ankit",
    "age": 21,
    "city": "Delhi"
}

print(student["name"])
print(student["age"])
```

## Using get() Method

The `get()` method returns the value of a key. If the key does not exist, it returns `None` (or a default value if provided).

```python
student = {
    "name": "Ankit",
    "age": 21
}

print(student.get("name"))
print(student.get("city"))
print(student.get("city", "Not Found"))
```

## Adding and Updating Items

New key-value pairs can be added, and existing values can be updated.

```python
student = {
    "name": "Ankit",
    "age": 21
}

student["city"] = "Delhi"
student["age"] = 22

print(student)
```

## Removing Items

Use `pop()`, `popitem()`, `del`, and `clear()` to remove items.

```python
student = {
    "name": "Ankit",
    "age": 21,
    "city": "Delhi"
}

student.pop("city")

print(student)
```

```python
student = {
    "name": "Ankit",
    "age": 21
}

student.popitem()

print(student)
```

```python
student = {
    "name": "Ankit",
    "age": 21
}

del student["age"]

print(student)
```

```python
student = {
    "name": "Ankit",
    "age": 21
}

student.clear()

print(student)
```

## Dictionary Methods

Some commonly used dictionary methods are:

- keys()
- values()
- items()

```python
student = {
    "name": "Ankit",
    "age": 21,
    "city": "Delhi"
}

print(student.keys())
print(student.values())
print(student.items())
```

## Looping Through a Dictionary

A `for` loop is used to iterate through dictionaries.

```python
student = {
    "name": "Ankit",
    "age": 21,
    "city": "Delhi"
}

for key in student:
    print(key)
```

```python
student = {
    "name": "Ankit",
    "age": 21,
    "city": "Delhi"
}

for value in student.values():
    print(value)
```

```python
student = {
    "name": "Ankit",
    "age": 21,
    "city": "Delhi"
}

for key, value in student.items():
    print(key, ":", value)
```

## Membership Operators

Use `in` and `not in` to check whether a key exists in a dictionary.

```python
student = {
    "name": "Ankit",
    "age": 21
}

print("name" in student)
print("city" in student)
print("city" not in student)
```

## Nested Dictionary

A dictionary can contain another dictionary.

```python
students = {
    "student1": {
        "name": "Ankit",
        "age": 21
    },
    "student2": {
        "name": "Rahul",
        "age": 22
    }
}

print(students)
```

# Chapter 20: Functions

## Definition

A **Function** is a reusable block of code that performs a specific task. Functions help reduce code repetition and make programs easier to read and maintain.

Python provides two types of functions:

1. Built-in Functions
2. User-defined Functions

```python
# Built-in Function

numbers = [10, 20, 30, 40, 50]

print(len(numbers))
print(max(numbers))
print(min(numbers))
print(sum(numbers))
```

## Creating a Function

A user-defined function is created using the `def` keyword.

### Syntax

def function_name():
    # Code

```python
def greet():
    print("Welcome to Python!")

greet()
```

## Function with Parameters

Parameters are variables that receive values when a function is called.

```python
def greet(name):
    print("Hello", name)

greet("Ankit")
greet("Rahul")
```

## Function with Multiple Parameters

```python
def add(a, b):
    print(a + b)

add(10, 20)
add(50, 40)
```

## Default Parameters

A default parameter has a predefined value. If no argument is passed, the default value is used.

```python
def greet(name="Guest"):
    print("Hello", name)

greet()
greet("Ankit")
```

## Keyword Arguments

Arguments can be passed using parameter names.

```python
def student(name, age):
    print(name)
    print(age)

student(age=21, name="Ankit")
```

## Variable-Length Arguments (*args)

Use `*args` when the number of arguments is unknown.

```python
def total(*numbers):
    print(sum(numbers))

total(10, 20, 30)
total(5, 10, 15, 20)
```

## Keyword Variable-Length Arguments (**kwargs)

Use `**kwargs` to accept multiple keyword arguments.

```python
def student_info(**details):
    print(details)

student_info(name="Ankit", age=21, city="Delhi")
```

## Lambda Function

A Lambda Function is a small anonymous function written in a single line.

```python
square = lambda x: x * x

print(square(5))
```

## Recursive Function

A Recursive Function is a function that calls itself.

```python
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```

## Local and Global Variables

- Local Variable: Created inside a function.
- Global Variable: Created outside a function.

```python
x = 100

def demo():
    y = 50
    print(x)
    print(y)

demo()

print(x)
```
